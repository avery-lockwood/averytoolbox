Find a plane containing the point $(-3,1,2)$ and the line of intersection of the planes $3x - y + 2z = -3$ and $2x + y - 3z = -40$.

````python
# The point
P = vector([-3,1,2])

# Normals of the given planes
n1 = vector([3,-1,2])
n2 = vector([2,1,-3])

# Direction vector of the intersection line
v = n1.cross_product(n2)

# Find a point on the intersection line by setting z=0 and solving for x, y
var('x y z')
sol = solve([3*x - y + 2*0 == -3, 2*x + y - 3*0 == -40], x, y)
if sol:
    Q = vector([sol[0][0].rhs(), sol[0][1].rhs(), 0])
else:
    # If z=0 doesn't work, try y=0
    sol = solve([3*x - 0 + 2*z == -3, 2*x + 0 - 3*z == -40], x, z)
    Q = vector([sol[0][0].rhs(), 0, sol[0][1].rhs()])

# The required plane contains P and the intersection line, so its normal is perpendicular to both v and (P-Q)
normal = v.cross_product(P - Q)

# Plane equation: a(x-x0) + b(y-y0) + c(z-z0) = 0
(a, b, c) = normal
x0, y0, z0 = P
plane = a*(x-x0) + b*(y-y0) + c*(z-z0)
plane = plane.expand()
show(plane == 0)
print('Plane equation:', plane)

# Also print in ax + by + cz = d form
#from sage.symbolic.relation import solve as ssolve
#plane_std = plane.expand().lhs() == 0
#print('Standard form:', plane_std)
````

Find the distance between the skew lines \( \vec{r}(t) = \langle 0,3,-1 \rangle t + \langle -7,-3,-8 \rangle \) and \( \vec{p}(s) = \langle -5,3,-3 \rangle s + \langle -4,2,4 \rangle \).

````python
# Define direction vectors and points on each line
v1 = vector([0,3,-1])
v2 = vector([-5,3,-3])
P1 = vector([-7,-3,-8])
P2 = vector([-4,2,4])

# The vector connecting any point on line 1 to any point on line 2
PQ = P2 - P1

# The distance between skew lines is the length of the projection of PQ onto the normal vector to both lines
n = v1.cross_product(v2)
distance = abs(PQ.dot_product(n)) / n.norm()
show(distance)
print('Distance between the skew lines:', distance)
````

Find the Tangent vector, the Normal vector, and the Binormal vector ($\vec{T}$, $\vec{N}$ and $\vec{B}$) for the curve $\vec{r}(t) = \langle 4\cos(3t), 4\sin(3t), 4t \rangle$ at the point $t = 0$.

````python
# Define variables and function
var('t')
r = vector([4*cos(3*t), 4*sin(3*t), 4*t])

# First derivative (velocity vector)
v = r.derivative(t)

# For Tangent vector: T = v/|v|
T = v / v.norm().simplify()

# Second derivative (acceleration vector)
a = v.derivative(t)

# For Normal vector: N = a - (a·T)T / |a - (a·T)T|
# First calculate a - (a·T)T
a_parallel = a.dot_product(T) * T
a_perp = a - a_parallel
N = a_perp / a_perp.norm().simplify()

# For Binormal vector: B = T × N
B = T.cross_product(N)

# Evaluate at t = 0
t_val = 0
T0 = T.subs(t=t_val)
N0 = N.subs(t=t_val)
B0 = B.subs(t=t_val)

print("At t = 0:")
print(f"T(0) = {T0}")
print(f"N(0) = {N0}")
print(f"B(0) = {B0}")

# Verify they are orthonormal
print("\nVerifying orthonormality:")
print(f"T·N = {T0.dot_product(N0).simplify()}")
print(f"T·B = {T0.dot_product(B0).simplify()}")
print(f"N·B = {N0.dot_product(B0).simplify()}")
print(f"|T| = {T0.norm().simplify()}")
print(f"|N| = {N0.norm().simplify()}")
print(f"|B| = {B0.norm().simplify()}")
````

Find the acute angle of intersection $\theta$ between the curves:
$\vec{r}_1(t) = \langle -3t, t^3, -2t^4 \rangle$ and 
$\vec{r}_2(t) = \langle \sin(-3t), \sin(-2t), t \rangle$ 
at the origin.

````python
# Define variables
var('t')

# Define the curves
r1 = vector([-3*t, t^3, -2*t^4])
r2 = vector([sin(-3*t), sin(-2*t), t])

# Find velocity vectors (tangent vectors) by taking derivatives
v1 = r1.derivative(t)
v2 = r2.derivative(t)

# Evaluate at t = 0 (the origin)
v1_0 = v1.subs(t=0)
v2_0 = v2.subs(t=0)

print("Velocity vectors at t=0:")
print(f"v1(0) = {v1_0}")
print(f"v2(0) = {v2_0}")

# Calculate angle using dot product formula: cos(θ) = (v1·v2)/(|v1||v2|)
dot_product = v1_0.dot_product(v2_0)
v1_mag = v1_0.norm()
v2_mag = v2_0.norm()
cos_theta = dot_product/(v1_mag * v2_mag)

# Get angle in radians
theta = arccos(cos_theta)

# Convert to degrees
theta_deg = theta * 180/pi

# Ensure we get the acute angle
if theta_deg > 90:
    theta_deg = 180 - theta_deg

print(f"\nAngle in radians: {theta}")
print(f"Angle in degrees: {theta_deg}")
````

Find the curvature of the curve $\vec{r}(t) = \langle 3\cos(2t), 3\sin(2t), 4t \rangle$ at the point $t = 0$ to two decimal places.

````python
# Define variables and the curve
var('t')
r = vector([3*cos(2*t), 3*sin(2*t), 4*t])

# First derivative r'(t)
r_prime = r.derivative(t)
print("r'(t) =", r_prime)

# Second derivative r''(t)
r_double_prime = r_prime.derivative(t)
print("r''(t) =", r_double_prime)

# Cross product r' × r''
cross_product = r_prime.cross_product(r_double_prime)

# Magnitude of cross product |r' × r''|
cross_product_mag = cross_product.norm()

# Magnitude of r' cubed |r'|³
r_prime_mag_cubed = r_prime.norm()^3

# Curvature κ = |r' × r''| / |r'|³
curvature = cross_product_mag / r_prime_mag_cubed

# Evaluate at t = 0
curvature_at_0 = curvature.subs(t=0).n(digits=10)

print(f"\nCurvature at t = 0: {curvature_at_0:.2f}")

# Show intermediate values at t = 0 for verification
print("\nAt t = 0:")
print(f"r'(0) = {r_prime.subs(t=0)}")
print(f"r''(0) = {r_double_prime.subs(t=0)}")
print(f"|r'(0) × r''(0)| = {cross_product_mag.subs(t=0).n()}")
print(f"|r'(0)|³ = {r_prime_mag_cubed.subs(t=0).n()}")
````

## Arc Length Function Usage

The `arc_length` function can be used in several ways:

1. Get the arc length function s(t):
```python
arc_length(vector([2*t, cos(t)]))  # 2D curve
arc_length(vector([2*t, cos(t), sin(t)]))  # 3D curve
```

2. Get the definite arc length over an interval:
```python
arc_length(vector([2*t, cos(t)]), a=0, b=pi)  # Compute length from t=0 to t=π
```

3. Get the arc length parameterization:
```python
arc_length(vector([2*t, cos(t)]), parameterize=True)  # Returns r(s)
```

````python
def arc_length(r, t, a=None, b=None, parameterize=False):
    """
    Calculate the arc length of a space curve or return its arc length parameterization.
    
    Parameters:
        r : A vector function r(t) in SageMath vector form
        a : Start parameter value (optional)
        b : End parameter value (optional)
        parameterize : If True, returns the arc length parameterization (default False)
    
    Returns:
        If parameterize=False and a,b=None: Returns arc length function s(t)
        If parameterize=False and a,b given: Returns definite arc length
        If parameterize=True: Returns r(s), the arc length parameterization
    """
    # Get the parameter variable (usually 't')
   # t = r.variables()[0]
    
    # Calculate dr/dt
    r_prime = np.array([diff(i) for i in r])
    
    # Calculate |dr/dt|
    speed = np.linalg.norm(r_prime)
    
    if parameterize:
        # For arc length parameterization, we need to:
        # 1. Find s(t) = ∫|r'(t)|dt
        s_of_t = integral(speed, t)
        
        # 2. Try to find t(s) by inverting s(t)
        try:
            # Attempt to solve s = s(t) for t
            t_of_s = solve(var('s') == s_of_t, t)[0].rhs()
            # Substitute t(s) back into r(t)
            var('s')
            return r.subs(t == t_of_s)
        except:
            print("Could not find explicit arc length parameterization")
            return None
    
    elif a is not None and b is not None:
        # Calculate definite arc length
        return integral(speed, t, a, b)
    
    else:
        # Return arc length function s(t)
        return integral(speed, t)

````

````python
def arc_length(r, t, parameterize=False, a=None, b=None):
    """
    Calculate the arc length of a space curve or return its arc length parameterization.

    Parameters:
        r : A SageMath vector function r(t)
        t : The parameter variable (e.g., t)
        a : Start parameter value (optional)
        b : End parameter value (optional)
        parameterize : If True, returns the arc length parameterization (default False)

    Returns:
        If parameterize=False and a,b=None: Returns arc length function s(t)
        If parameterize=False and a,b given: Returns definite arc length
        If parameterize=True: Returns r(s), the arc length parameterization(
    """
    # Compute the derivative vector r'(t)
    r_prime = np.array([diff(i) for i in r])
    # Compute the speed |r'(t)|
    speed = sqrt(sum([i^2 for i in r_prime]))

    # Arc length function s(t) = ∫_{t0}^{t} |r'(u)| du, usually from t0=0
    if parameterize:
        # 1. Find s(t)
        s_of_t = integral(speed, t)
        # 2. Solve for t in terms of s: s = s(t) => t = t(s)
        try:
            s = var('s')
            t_of_s = solve(s == s_of_t, t)[0].rhs()
            # 3. Substitute t(s) into r(t)
            r_s = r.subs(t=t_of_s)
            return r_s
        except Exception as e:
            print("Could not find explicit arc length parameterization:", e)
            return None
    elif a is not None and b is not None:
        # Definite arc length from a to b
        return integral(speed, t, a, b).full_simplify()
    else:
        # Indefinite arc length function s(t)
        return integral(speed, t).full_simplify()
````

````python
def to_cartesian(coords):
    """
    Converts polar (2D) or spherical (3D) coordinates to Cartesian coordinates.
    - 2D: (r, theta) → (x, y)
    - 3D: (rho, theta, phi) → (x, y, z)
    Angles should be in radians.
    """
    if len(coords) == 2:
        r, theta = coords
        x = r * cos(theta)
        y = r * sin(theta)
        return (x, y)
    elif len(coords) == 3:
        rho, theta, phi = coords
        x = rho * sin(phi) * cos(theta)
        y = rho * sin(phi) * sin(theta)
        z = rho * cos(phi)
        return (x, y, z)
    else:
        raise ValueError("Input must be 2D (r, theta) or 3D (rho, theta, phi)")
````
